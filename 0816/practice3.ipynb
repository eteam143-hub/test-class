```python
print(3 + 5)          # 加法 → 8
print(10 - 4)         # 減法 → 6
print(3 * 5)          # 乘法 → 15
print(15 / 4)         # 除法 → 3.75（永遠回傳浮點數）
print(15 // 4)        # 整數除法 → 3（只取商的整數部分）
print(15 % 4)         # 餘數運算 → 3（15 ÷ 4 的餘數）
print(2 ** 10)        # 次方運算 → 1024（2 的 10 次方）
```

```python
print(3 + 5 * 2)      # 先乘後加 → 13
print((3 + 5) * 2)    # 先加再乘 → 16
```

```python
number = 17
if number % 2 == 0:
    print(number, "是偶數")
else:
    print(number, "是奇數")
# 輸出：17 是奇數
```

```python
import math

x1, y1 = 1, 2
x2, y2 = 4, 6
distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
print("兩點距離 =", distance)   # 輸出：兩點距離 = 5.0
```
